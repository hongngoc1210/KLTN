```python
from pathlib import Path

import py_vncorenlp


VNCORENLP_DIR = Path(
    "/home/trang/study/khoaluan/KLTN/src/qa_models/resources/vncorenlp"
)

VNCORENLP_DIR.mkdir(parents=True, exist_ok=True)

jar_path = VNCORENLP_DIR / "VnCoreNLP-1.2.jar"

# Chỉ tải một lần nếu chưa có model.
if not jar_path.exists():
    py_vncorenlp.download_model(
        save_dir=str(VNCORENLP_DIR),
    )
    print(f"Downloaded VnCoreNLP to: {VNCORENLP_DIR}")
else:
    print(f"VnCoreNLP already exists at: {VNCORENLP_DIR}")
```

```python
segmenter = py_vncorenlp.VnCoreNLP(
    annotators=["wseg"],
    save_dir=str(VNCORENLP_DIR),
)

text = (
    "Ông Nguyễn Khắc Chúc đang làm việc tại Đại học Quốc gia Hà Nội. "
    "Bà Lan, vợ ông Chúc, cũng làm việc tại đây."
)

segmented_sentences = segmenter.word_segment(text)

for sentence in segmented_sentences:
    print(sentence)
```
